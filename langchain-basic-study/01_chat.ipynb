```python
import os

from dotenv import load_dotenv

load_dotenv()

if not os.environ.get("OPENAI_API_KEY"):
    raise SystemExit(
        "❌ OPENAI_API_KEY가 설정되어 있지 않습니다.\n"
        "   .env 파일에 OPENAI_API_KEY=<발급받은 키> 를 추가한 뒤 커널을 재시작하세요."
    )
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-4o-mini") # OPENAI_API_KEY

llm.invoke("너는 누구냐?")
```

```python
from langchain_openai import AzureChatOpenAI

# Azure OpenAI는 .env에 아래 값 필요:
# AZURE_OPENAI_API_KEY, AZURE_OPENAI_ENDPOINT, AZURE_OPENAI_API_VERSION, AZURE_OPENAI_DEPLOYMENT
missing_azure_vars = [
    name
    for name in ("AZURE_OPENAI_API_KEY", "AZURE_OPENAI_ENDPOINT")
    if not os.environ.get(name)
]
if missing_azure_vars:
    raise SystemExit(
        f"❌ Azure OpenAI 필수 환경변수가 없습니다: {', '.join(missing_azure_vars)}\n"
        "   .env 파일에 해당 값을 채운 뒤 커널을 재시작하세요."
    )

llm = AzureChatOpenAI(
    azure_endpoint=os.environ.get("AZURE_OPENAI_ENDPOINT"),
    api_version=os.environ.get("AZURE_OPENAI_API_VERSION", "2024-06-01"),
    azure_deployment=os.environ.get("AZURE_OPENAI_DEPLOYMENT", "gpt-4o-mini"),
)

llm.invoke("너는 누구냐?")
```
